```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

**Problem statement**
The business faces suboptimal purchase conversion rates, as indicated by a significantly larger number of non-purchasers compared to purchasers.  This necessitates understanding the key factors influencing purchase decisions to improve marketing effectiveness and increase conversions.  The business needs to determine how demographics (gender, age) and financial indicators (estimated salary) relate to purchase behavior to identify target customer segments and develop tailored marketing strategies that resonate with these groups, ultimately driving higher purchase rates."

```python
df = pd.read_csv("Social_Network_Ads.csv")
df
```

```python
df['Purchased'] = df['Purchased'].replace({0: 'No', 1: 'Yes'})
```

The distribution of purchase behavior (purchased/did not purchase) 

```python
plt.figure(figsize=(7,5))
sns.countplot(x='Purchased',data=df,palette="pastel")
plt.xlabel("purchase",fontsize=10)
plt.ylabel("count")
plt.title("Distribution of Customer who purchased")
plt.show()
```

The graph shows a significantly higher number of "No" purchases compared to "Yes" purchases, indicating a potential area for improvement in converting non-purchasers.

Distribution of customers Gender

```python
plt.figure(figsize=(7,5))
sns.countplot(x="Gender",data=df,palette="pastel")
plt.xlabel("Gender",fontsize=10)
plt.ylabel("count")
plt.title("Distribution of customers Gender")
plt.show()
```

```python
df.Gender.value_counts()
```

Distribution of Gender wise purchase

```python
plt.figure(figsize=(7,5))
sns.countplot(x="Gender",hue='Purchased',data=df,palette="pastel")
plt.xlabel("Gender",fontsize=10)
plt.ylabel("count")
plt.title("Distribution of Gender wise purchase")
plt.show()
```

The chart reveals a gender-based purchase pattern: females show a higher tendency to purchase compared to males, while males have a larger gap between non-purchasers and purchasers. This suggests that marketing strategies could be tailored to better appeal to male potential customers.

Age wise purchase

```python
sns.boxplot(hue='Purchased', x='Age', data=df)
plt.title('Age wise purchase')
plt.show()
```

The boxplot indicates that the age group that tends to make purchases skews slightly older, with the median age of purchasers appearing to be in the late 40s or early 50s.  While there's overlap, the overall distribution for purchasers is shifted towards older ages compared to non-purchasers.

Distribution of Age and Gender wise purchase

```python
sns.boxplot(y='Age', x='Gender',data=df,hue='Purchased')
plt.title('Distribution of Age and Gender wise purchase')
plt.show()
```

he boxplot compares the estimated salary distribution across gender, segmented by purchase status. It reveals that, while there's overlap, purchasers, especially females, tend to have slightly higher estimated salaries on average, suggesting income level could influence purchase decisions.

Gender wise Salary

```python
sns.boxplot(y='EstimatedSalary', x='Gender',data=df)
plt.title('Gender wise Salary')
plt.show()
```

There appears to be a slight difference in estimated salaries between genders.  Visually, the median salary for females seems slightly higher than the median salary for males. 

Gender and Salary wise purchase

```python

sns.boxplot(y='EstimatedSalary', x='Gender',data=df,hue='Purchased')
plt.title('Gender and Salary wise purchase')
plt.show()
```

For both males and females, those who purchased the product ("Yes") tend to have higher estimated salaries compared to those who didn't ("No"). This suggests that individuals with higher salaries are more likely to make a purchase.

Specifically, females who made a purchase appear to have notably higher median salaries than males who made a purchase.

**conclusion**


Exploratory data analysis revealed key insights into customer purchase behavior.  Overall, the number of non-purchasers significantly outweighs purchasers, highlighting a need for improved conversion strategies.  Purchase patterns exhibit gender-based differences, with females demonstrating a higher purchase tendency and males showing a larger gap between non-purchasers and purchasers, suggesting targeted marketing opportunities.  Age also appears to be a factor, as purchasers skew slightly older, typically in their late 40s or early 50s.  Furthermore, salary plays a role, with purchasers, especially females, generally having higher estimated salaries, indicating income level's influence on purchase decisions.
